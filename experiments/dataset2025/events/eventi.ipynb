```python
import pandas as pd

# Lista dei mesi da generare
mesi = ["2025-03", "2025-05", "2025-08", "2025-09"]

# Lista vuota per accumulare le date
tutte_date = []

for mese in mesi:
    # Crea un range di date per il mese
    start = f"{mese}-01"
    end = pd.Timestamp(start) + pd.offsets.MonthEnd(0)
    tutte_date.extend(pd.date_range(start, end))

# Crea il DataFrame
df_eventi = pd.DataFrame({
    "day": tutte_date,
    "evento_dozza": 0,
    "evento_imola" : 0,
    "evento_italia" : 0
})

# Liste delle date in cui evento = 1
date_evento_dozza = ["2025-03-02", "2025-03-09", "2025-03-23",
                     "2025-09-05", "2025-09-06", "2025-09-07", "2025-09-08", "2025-09-14", "2025-09-21"]
date_evento_imola = ["2025-03-30", "2025-05-17", "2025-05-18"]
date_evento_italia = ["2025-05-01", "2025-05-02", "2025-05-03", "2025-08-15"]


# Converti in datetime
date_evento_dozza = pd.to_datetime(date_evento_dozza)
date_evento_imola = pd.to_datetime(date_evento_imola)
date_evento_italia = pd.to_datetime(date_evento_italia)

# Imposta 1 per le date specifiche
df_eventi.loc[df_eventi["day"].isin(date_evento_dozza), "evento_dozza"] = 1
df_eventi.loc[df_eventi["day"].isin(date_evento_imola), "evento_imola"] = 1
df_eventi.loc[df_eventi["day"].isin(date_evento_italia), "evento_italia"] = 1

# Salva in CSV
df_eventi.to_csv("eventi_dozza_imola_italia.csv", index=False)

df_eventi
```
